````python
#Problem 2 (3)使用 Python 编程实现上述算法或其对应的更高效版本的算法
# 如果允许写更高效版本，则可以使用归并排序，其时间复杂度为 O(nlogn)
import math

def merge_sort(A):
    if len(A) <= 1:
        return A

    mid = len(A) // 2
    left = merge_sort(A[:mid])
    right = merge_sort(A[mid:])

    return merge(left, right)

def merge(left, right):
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])
    return result


# 测试
arr = [5, 2, 8, 1, 4, 7, 3, 6]
print("原数组:", arr)
print("归并排序后:", merge_sort(arr))
````

## 问题 3

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

## 解答

### 1. 设计一个算法计算最大利润和对应买入卖出日期

由于 `profits[i]` 表示第 \(i\) 天的收益变化，因此从第 \(i\) 天买入、第 \(j\) 天卖出（\(i<j\)）的总利润等于区间和：

$$
profits[i] + profits[i+1] + \cdots + profits[j-1]
$$

所以问题可以转化为：在数组中找到和最大的连续子数组，并输出对应的起止位置。

最直接的方法是枚举所有买入日和卖出日，计算每个区间的收益，并记录最大值。

时间复杂度为：

$$
O(n^2)
$$

空间复杂度为：

$$
O(1)
$$

Python 实现如下：

```python


````python
def max_profit_bruteforce(profits):
    n = len(profits)
    max_profit = float('-inf')
    buy_day = -1
    sell_day = -1

    for i in range(n):
        current_sum = 0
        for j in range(i, n):
            current_sum += profits[j]
            if current_sum > max_profit:
                max_profit = current_sum
                buy_day = i + 1          # 第几天买入（从1开始记）
                sell_day = j + 2         # 第几天卖出

    return max_profit, buy_day, sell_day


profits = [3, 2, 1, -7, 5, 2, -1, 3, -1]
print("profits=",profits)
ans1 = max_profit_bruteforce(profits)
print("暴力解时间复杂度是 O(n^2) ，暴力解answer:", ans1)
````

### 2. 设计一个时间复杂度为 $O(n)$ 的算法

#### 思路

可以使用 **Kadane 算法** 来解决最大子数组和问题。

定义：

- `current_sum`：当前连续子数组的和  
- `max_profit`：到目前为止找到的最大利润  

在遍历数组时：

- 如果当前累计和小于 0，那么继续累加只会降低后续结果，因此应从当前位置重新开始；
- 否则，将当前元素加入已有子数组；
- 在每一步更新最大利润时，同时记录对应的买入日和卖出日。

#### 算法过程

遍历数组一次，对于每个位置 $i$：

- 更新当前子数组和：
  
$$
current\_sum = \max(profits[i],\ current\_sum + profits[i])
$$

- 更新最大利润：

$$
max\_profit = \max(max\_profit,\ current\_sum)
$$

#### 复杂度分析

- 时间复杂度：

$$
O(n)
$$

- 空间复杂度：

$$
O(1)
$$
Python 实现如下：

```python


````python
def max_profit_linear(profits):
    best_sum = float('-inf')
    cur_sum = 0
    start = 0
    best_l = best_r = 0

    for i, x in enumerate(profits):
        if cur_sum <= 0:
            cur_sum = x
            start = i
        else:
            cur_sum += x

        if cur_sum > best_sum:
            best_sum = cur_sum
            best_l = start
            best_r = i

    return best_sum, best_l + 1, best_r + 2


profits = [3, 2, 1, -7, 5, 2, -1, 3, -1]
print("profits =", profits)
print("此算法时间复杂度为O(n)   answer:", max_profit_linear(profits))
````

## 问题 4

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为“局部最小”。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

## 解答

### 1. 找到一个局部最小值

由于题目满足：

$$
A[1] > A[2], \quad A[n-1] < A[n]
$$

因此数组左侧先下降、右侧最终上升，所以中间一定至少存在一个局部最小值。

可以使用二分查找：

- 若中间元素同时小于左右邻居，则它就是局部最小；
- 若中间元素大于左邻居，则局部最小一定在左半边；
- 否则局部最小一定在右半边。

因此每次都能将搜索区间缩小一半，时间复杂度为：

$$
O(\log n)
$$

Python 实现如下：


````python
def find_one_local_min(A):
    left = 1
    right = len(A) - 2   # 只在内部位置查找，因为局部最小要求有左右邻居

    while left <= right:
        mid = (left + right) // 2

        if A[mid] < A[mid - 1] and A[mid] < A[mid + 1]:
            return A[mid]

        elif A[mid] > A[mid - 1]:
            right = mid - 1

        else:
            left = mid + 1

    return None
A = [9, 3, 7, 2, 1, 4, 5]
print("数组:", A)
print("找到的一个局部最小值:", find_one_local_min(A))
````

### 2. 找出所有局部最小值

#### 思路

如果要求找出所有局部最小值，可以直接遍历整个数组。

对于每一个内部位置 $i$（即 $1 \le i \le n-2$），只需要判断该元素是否同时小于左右邻居：

$$
A[i] < A[i-1] \quad \text{且} \quad A[i] < A[i+1]
$$

如果满足上述条件，则 $A[i]$ 为一个局部最小值，将其加入结果中。

由于需要检查数组中的每一个元素，因此该算法的时间复杂度不可能优于线性时间。

#### 时间复杂度

$$
O(n)
$$

#### 空间复杂度

$$
O(k)
$$

其中 $k$ 为局部最小值的个数（用于存储结果）。如果不计输出空间，则额外空间复杂度为：

$$
O(1)
$$

#### Python 实现

````python
def find_all_local_mins(A):
    result = []
    for i in range(1, len(A) - 1):
        if A[i] < A[i - 1] and A[i] < A[i + 1]:
            result.append(A[i])
    return result
A = [9, 3, 7, 2, 1, 4, 5]
print("数组:", A)
print("所有局部最小值:", find_all_local_mins(A))
````

## 问题 5

给定包含n个不同数字的一组数, 寻找一种基于比较的算法在这组数中找到k个最小的数字, 并按顺序输出它们。

3. \* 假设数组中包含的数据总数目超过了计算机的存储能力, 请设计一个算法, 找到这堆数据的前k小的数值, 计算时间复杂度, 并使用python实现该算法, 假设计算机一定能存储k个数据。

e.g.：
---
数组arr=[5,4,3,2,6,1,88,33,22,107] 的前3个最小数据为：1, 2, 3

### 解答

### 1. 先排序，再输出最小的 k 个数

这是最直接的方法。先对整个数组排序，再取前 $k$ 个元素即可。

如果使用比较排序算法，则排序时间复杂度为：

$$
O(n \log n)
$$

输出前 $k$ 个元素还需要：

$$
O(k)
$$

因此总时间复杂度为：

$$
O(n \log n + k)
$$

由于一般 k ≤ n，所以通常也可以写成：

$$
O(n \log n)
$$

Python 实现如下：


````python
def k_smallest_by_sort(arr, k):
    sorted_arr = sorted(arr)
    return sorted_arr[:k]


arr = [12, 5, 7, 1, 19, 3, 8, 2, 10]
k = 4

print("原数组:", arr)
print("最小的 k 个数:", k_smallest_by_sort(arr, k))
````

### 2. 建立一个包含这 $n$ 个数的堆，并调用 $k$ 次 Extract-min

#### 思路

题目要求使用“向空堆中不断插入元素”的方式建堆。假设我们使用最小堆：

- 先把数组中的 $n$ 个元素逐个插入堆中；
- 然后执行 $k$ 次 `Extract-min`，每次取出当前最小值；
- 这样取出的 $k$ 个元素就是按顺序排列的最小 $k$ 个数。


#### 时间复杂度分析
##### 建堆阶段
若从空堆开始，逐个插入 $n$ 个元素：
- 每次插入最坏需要：

$$
O(\log n)
$$

- 因此建堆总时间复杂度为：

$$
O(n \log n)
$$

##### 提取阶段
执行 $k$ 次 `Extract-min`：
- 每次操作复杂度为：

$$
O(\log n)
$$

- 因此总复杂度为：

$$
O(k \log n)
$$

##### 总复杂度
因此总时间复杂度为：

$$
O(n \log n + k \log n)
$$

也可以写成：

$$
O((n + k)\log n)
$$
Python 实现如下：

````python
import heapq

def k_smallest_by_heap(arr, k):
    heap = []
    for x in arr:
        heapq.heappush(heap, x)

    result = []
    for _ in range(k):
        result.append(heapq.heappop(heap))

    return result


arr = [12, 5, 7, 1, 19, 3, 8, 2, 10]
k = 4

print("原数组:", arr)
print("最小的 k 个数:", k_smallest_by_heap(arr, k))
````

### 3. 数据量超过内存时，如何处理前 $k$ 小的数值
#### 思路
当数据量非常大、无法全部装入内存时，就不能对整体数据进行排序，也不适合把全部元素都建成一个堆。

题目说明：假设计算机一定能存储 $k$ 个数据。
因此，可以维护一个**大小为 $k$ 的最大堆**，具体做法如下：

1. 先读取前 $k$ 个元素，建立一个大小为 $k$ 的最大堆；
2. 对于后续读到的每一个新元素：
   - 如果它大于等于堆顶，则说明它不可能进入前 $k$ 小，直接丢弃；
   - 如果它小于堆顶，则说明它应该进入前 $k$ 小，此时替换堆顶并重新调整堆；
3. 遍历完全部数据后，堆中保留的就是最小的 $k$ 个数；
4. 最后将这 $k$ 个数排序输出。

由于堆顶始终是当前这 $k$ 个数中最大的元素，因此可以高效维护“前 $k$ 小”。

#### 时间复杂度分析
对于每一个元素，最多进行一次堆调整，其复杂度为：

$$
O(\log k)
$$

共有 $n$ 个元素，因此总时间复杂度为：

$$
O(n \log k)
$$

最后对这 $k$ 个元素进行排序输出，需要：

$$
O(k \log k)
$$

因此总时间复杂度为：

$$
O(n \log k + k \log k)
$$

如果只关注主导项，可以写为：

$$
O(n \log k)
$$


#### 空间复杂度
只需要维护一个大小为 $k$ 的堆，因此空间复杂度为：

$$
O(k)
$$
Python 实现如下：

````python
import heapq

def k_smallest_stream(arr, k):
    max_heap = []

    for x in arr:
        if len(max_heap) < k:
            heapq.heappush(max_heap, -x)
        else:
            if x < -max_heap[0]:
                heapq.heapreplace(max_heap, -x)

    result = sorted([-x for x in max_heap])
    return result


arr = [12, 5, 7, 1, 19, 3, 8, 2, 10]
k = 4

print("原数组:", arr)
print("最小的 k 个数:", k_smallest_stream(arr, k))
````

## 问题 6

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

### 解答

可以使用一个栈来模拟当前还存活的星球。

只有一种情况会发生碰撞：  
**左边星球向右飞（正数），右边星球向左飞（负数）**。

处理当前星球 `x` 时：
1. 如果栈顶星球质量更小，则栈顶消失，继续比较
2. 如果栈顶星球质量更大，则当前星球消失
3. 如果质量相同，根据题目要求：  
   **向左运动的星球消失，向右运动的星球保留**  
   所以当前负数星球消失，栈顶留下

每个星球最多进栈一次、出栈一次，因此总时间复杂度为 \(O(n)\)。


````python
def asteroid_collision(A):
    stack = []

    for x in A:
        alive = True
        while alive and stack and stack[-1] > 0 and x < 0:
            if stack[-1] < -x:
                stack.pop()
            elif stack[-1] == -x:
                alive = False    # 题目要求：向左运动的消失，向右运动的保留
            else:
                alive = False

        if alive:
            stack.append(x)

    return stack

case1 = [-3, -6, 2, 8, 5, -8, 9, -2, 1]
case2 = [23, -8, 9, -3, -7, 9, -23, 22]
print("case1 input :", case1)
print("case1 output:", asteroid_collision(case1))
print("case2 input :", case2)
print("case2 output:", asteroid_collision(case2))
print("time complexity: O(n)")
````

## 问题 7

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

### 解答

这是经典的 **0-1 背包问题**。  
设 `dp[i][w]` 表示：前 `i` 个物品中，在容量不超过 `w` 时能获得的最大价值。

对于第 `i` 个物品（重量为 `weight`，价值为 `value`），有两种选择：
- 不选它：`dp[i-1][w]`
- 选它（前提是放得下）：`dp[i-1][w-weight] + value`

因此状态转移方程为
\[
dp[i][w]=\max\{dp[i-1][w],\ dp[i-1][w-weight_i]+value_i\}
\]

本题中物品为 `[(5,10), (4,40), (6,30), (3,50)]`，背包容量为 9。  
二维 DP 的时间复杂度为 \(O(nW)\)，空间复杂度为 \(O(nW)\)。


````python
def knapsack_01(items, W):
    n = len(items)
    dp = [[0] * (W + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        weight, value = items[i - 1]
        for w in range(W + 1):
            dp[i][w] = dp[i - 1][w]
            if weight <= w:
                dp[i][w] = max(dp[i][w], dp[i - 1][w - weight] + value)

    return dp[n][W]

items = [(5,10), (4,40), (6,30), (3,50)]
W = 9
print("items =", items)
print("capacity =", W)
print("maximum value =", knapsack_01(items, W))
print("time complexity: O(nW)")
````

给你二叉搜索树的根节点 root ，该树中的恰好两个节点的值被错误地交换。请在不改变其结构的情况下，恢复这棵树 。设计算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

输入： root=[3,1,4,null,null,2]， 输出为：[2,1,4,null,null,3]

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

### 解答

二叉搜索树的中序遍历结果应该是**严格递增**的。  
如果有两个节点的值被交换，那么中序遍历序列中会出现逆序对。

做法：
1. 进行一次中序遍历
2. 用 `pred` 记录前一个访问到的节点
3. 如果发现 `pred.val > cur.val`，说明出现逆序
   - 第一次逆序时记录 `x = pred`
   - 每次逆序都更新 `y = cur`
4. 最后交换 `x.val` 和 `y.val`

这样不改变树结构，只交换两个错误节点的值即可恢复 BST。

时间复杂度为 \(O(n)\)。  
若使用显式栈做中序遍历，额外空间复杂度为 \(O(h)\)，其中 \(h\) 是树高。


````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def recover_tree(root):
    stack = []
    x = y = pred = None
    cur = root

    while stack or cur:
        while cur:
            stack.append(cur)
            cur = cur.left

        cur = stack.pop()

        if pred and pred.val > cur.val:
            y = cur
            if x is None:
                x = pred

        pred = cur
        cur = cur.right

    if x and y:
        x.val, y.val = y.val, x.val

def inorder(root):
    if not root:
        return []
    return inorder(root.left) + [root.val] + inorder(root.right)

# 示例：root = [1,3,null,null,2]
root = TreeNode(1)
root.left = TreeNode(3)
root.left.right = TreeNode(2)

print("before recover:", inorder(root))
recover_tree(root)
print("after recover :", inorder(root))
print("time complexity: O(n), extra space: O(h)")
````

## 问题 9 \*
给定 `n` 个非负整数表示每个宽度为 `1` 的柱子的高度图，计算按此排列的柱子，下雨之后能接多少雨水。

### 解答

这是经典的接雨水问题。  
使用双指针方法：

- `left` 从左往右扫
- `right` 从右往左扫
- `left_max` 记录左侧最高柱子
- `right_max` 记录右侧最高柱子

如果 `height[left] < height[right]`，那么当前位置左边能接多少水只取决于 `left_max`；  
否则当前位置右边能接多少水只取决于 `right_max`。

这样每个位置只访问一次，因此：
- 时间复杂度：\(O(n)\)
- 空间复杂度：\(O(1)\)


````python
def trap(height):
    left, right = 0, len(height) - 1
    left_max = right_max = 0
    water = 0

    while left < right:
        if height[left] < height[right]:
            if height[left] >= left_max:
                left_max = height[left]
            else:
                water += left_max - height[left]
            left += 1
        else:
            if height[right] >= right_max:
                right_max = height[right]
            else:
                water += right_max - height[right]
            right -= 1

    return water

height = [0,1,0,2,1,0,1,3,2,1,2,1]
print("height =", height)
print("trapped water =", trap(height))
print("time complexity: O(n), space complexity: O(1)")
````

## 问题 10

给你一个二叉树的根节点 root ，树中每个节点都存放有一个 0 到 9 之间的数字。每条从根节点到叶节点的路径都代表一个数字：

> 例如，从根节点到叶节点的路径 1 -> 2 -> 3 表示数字 123 。

设计一个算法计算从根节点到叶节点生成的所有数字之和 ，分析算法设计思路，计算时间复杂度，并基于python编程实现

叶节点:是指没有子节点的节点。

e.g.:
---
1.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-1.jpg"> </div>

 输入：root = [1,2,3], 输出：25

解释：从根到叶子节点路径 1->2 代表数字 12，1->3 代表数字 13，因此和为 12 + 13 = 25 。

2.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-2.jpg"> </div>

 输入：root = [4,9,0,5,1], 输出：1026

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

### 解答

从根到叶子的一条路径可以看成一个十进制数字。  
例如路径 `1 -> 2 -> 3` 表示数字 123。

因此可以使用 DFS：
- 当前走到某个节点时，把之前形成的数字乘 10 再加上当前节点值
- 如果走到叶子节点，就把这个完整数字返回
- 否则递归计算左右子树的结果并相加

整棵树中每个节点只访问一次，因此时间复杂度为 \(O(n)\)。  
递归栈深度为树高，所以额外空间复杂度为 \(O(h)\)。


````python
class TreeNode2:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def sum_numbers(root):
    def dfs(node, cur):
        if not node:
            return 0

        cur = cur * 10 + node.val

        if not node.left and not node.right:
            return cur

        return dfs(node.left, cur) + dfs(node.right, cur)

    return dfs(root, 0)

# 示例：root = [1,2,3]
root = TreeNode2(1, TreeNode2(2), TreeNode2(3))
print("sum of root-to-leaf numbers =", sum_numbers(root))
print("time complexity: O(n), extra space: O(h)")
````
